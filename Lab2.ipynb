**Note 1**: in order to have a similar shape, we need to select the amplitudes and phases carefully. One way to do this is to define the harmonic signal as follows:

$$y(t) = \sum_{k=1}^K A_k\cos\left(2\pi kf_0 t + k \phi - (k-1)\pi/2 \right), $$

where $K$ is the number of harmonics, $f_0$ is the fundamental frequency, $A_k$ is the amplitude (weight) of each harmonic and $\phi$ is the phase of the signal (defined in Lab1 Ex 2.4).

**Note 2**:
Define the $A_k$ values relative to the fundamental frequency. This means to define $A_1=1$ and the others less than 1. You can use Audacity to plot the spectrum of the fragment selected of the reference audio and measure the relative amplitudes of the harmonics.

**Note 3**: Normalize the amplitude of the signal by the same amplitude of the reference. For instance, if the amplitude of the reference signal is 0.33, you can normalize the syntesized signal by first dividing by its maximum and then multiplyng by 0.33:

### Showing the frequency spectrum of our reference sound

With the following code you can check the spectrogram

````python
# Write the code here
def plt_spectrum(z, t_start=0.763):
    windows_len = 8192
    ff, tt, S = signal.spectrogram(z, 44100, nperseg=windows_len, noverlap=windows_len/2)
    print(tt)

    plot_spectrum_at(ff, tt, S, t_start)

plt_spectrum(ref, 0.5)
````

In this code we used the spectrogram and the values that we got from Lab 1 to try and reconstruct the signal using the fourier series. In the following example we used the initial phase to compute phases for each k harmonic as shown in the previous equation.

````python
windows_len = 8192
ff, tt, S = signal.spectrogram(ref, 44100, nperseg=windows_len, noverlap=windows_len/2)

t_start = 0
t_hop =  np.argmin((tt - t_start)**2)
ftt = S[:, t_hop]

t_out = np.arange(len(ref)) / fs
x_total = np.zeros_like(t_out)

first_harmonic = 416.037

initial_phase = ((0.317277 - 0.31608)/0.0024266) * 2 * np.pi

for i in range(1,100):
    k_harmonic = i * first_harmonic
    f_hop = np.argmin((ff - k_harmonic)**2)

    amplitude = np.sqrt(ftt[f_hop])*2

    phase = i*initial_phase - (i-1) * np.pi/2

    x_total += amplitude * np.cos(2*np.pi*k_harmonic*t_out + phase)

plot_signals([x_total, ref], fs, t_start=tt[t_hop], t_end=tt[t_hop+1])
````

Complications that we saw with this method is that first of all if we take a better look at the spectrogram we can see that the harmonics are not perfect, in the sense that they don't exactly follow the the frequency principle (fk = f0*k). Furthermore we don't know the exact phase that these harmonics use. For these reasons we decided to use the function `signal.stft()` to compute both the amplitude and the phase at each frequency.

````python
windows_len = 8192

f, t_spec, Zxx = signal.stft(ref, 44100, nperseg=windows_len, noverlap=windows_len/2)

amplitude = np.abs(Zxx)
phase = np.angle(Zxx)

print()

plot_spectrum_at(f, t_spec, [amplitude, phase], 0.1)
````

First we solved the problem with the phase difference to show that even with the correct amplitude and phase of the harmonics we don't get such a good result.

````python
windows_len = 8192
f, t_spec, Zxx = signal.stft(ref, 44100, nperseg=windows_len, noverlap=windows_len/2)

t_start = 0.1
t_hop =  np.argmin((t_spec - t_start)**2)
ftt = Zxx[: , t_hop]

t_out = np.arange(len(ref)) / fs
x_total = np.zeros_like(t_out)

first_harmonic = 416.037

for i in range(10):
    k_harmonic = (i + 1) * first_harmonic
    f_hop = np.argmin((f - k_harmonic)**2)

    amplitude = np.abs(ftt[f_hop])*2
    phase = np.angle(ftt[f_hop])

    x_total += amplitude * np.cos(2*np.pi*k_harmonic*t_out + phase)

print(t_spec[t_hop], t_spec[t_hop+1])

plot_signals([x_total, ref], fs, t_start=t_spec[t_hop], t_end=t_spec[t_hop+1])
````

Now if we manually picked the frequencies of the sound that had the biggest amplitude we get a signal closer to the original one.

````python
windows_len = 8192
f, t_spec, Zxx = signal.stft(ref, 44100, nperseg=windows_len, noverlap=windows_len/2)

t_start = 0.1
t_hop =  np.argmin((t_spec - t_start)**2) 
ftt = Zxx[: , t_hop]

t_out = np.arange(len(ref)) / fs
x_total = np.zeros_like(t_out)

harmonics = [415, 824, 1244, 1669, 2095, 2536]

for frequency in harmonics:
    f_hop = np.argmin((f - frequency)**2)

    amplitude = np.abs(ftt[f_hop])*2
    phase = np.angle(ftt[f_hop])

    x_total += amplitude * np.cos(2*np.pi*frequency*t_out + phase)

print(t_spec[t_hop], t_spec[t_hop+1])

plot_signals([x_total, ref], fs, t_start=t_spec[t_hop], t_end=t_spec[t_hop+1])
````
